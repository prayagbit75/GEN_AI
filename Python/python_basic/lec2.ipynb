# list Comprehension 

### we are using looping statement inside the list for store the value inside the list in an iterable formate and in list comprehension constional statement is allowed 

```python
# inster 1 t 100 number inside the list 

nums = [x for x in range(1 , 101)]
print(nums)
```

```python
# i want to store square of even number inside the list 
even_sqr = [ x**2 for x in range(1,51) if x%2==0]
print(even_sqr)
```

## i can use funtion also in list comprehension

```python
# eg i want to make a list in which store len of string that are present in my girls list 

girls = ['riya' , 'siya' , 'aru' , 'yatika']

girls_len = [ len(name) for name in girls]
print(girls_len)
```

## nested for loop can be use 

```python
pair = [[i,j] for i in girls for j in girls_len]
print(pair)
```
