# Recursion

### By recursion function we can say that a function is going to call itself again and again 

```python
# factorial

def fact(num):
    if num==0 or num==1:
        return 1

    return num*fact(num-1)    
```

```python
print(fact(4))
```

## function with variable lenght argument

#### positional argument : positional argument are used so that we can pass multiple values inside the function as an argument  

```python
def total(*num):
    ans = 0
    for i in num:
        ans += i
    return ans    
```

```python
# count total line , total words , total char


with open('example.txt','r') as f:
    lines = f.readlines()             # it returns the list of lines 
    print(f"total line : {len(lines)}")
    total_words = sum(len(line.split())  for line in lines)
    print(f"total words : {total_words}")
    total_char = sum(len(line) for line in lines)
    print(f"total char : {total_char}")
```
