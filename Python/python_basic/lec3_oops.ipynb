## Inheritance 
#### Inheritance means child class in herit the attribute and funtionality from parent class is called inheritance 

```python
# Single level inheritance A single child class inherits from a single parent class.

class parent :
    def  __init__(self,name):
        self.name = name

    def getname(self):
        print(self.name)


class child(parent):

    def  __init__(self,name,age):
        super().__init__(name)
        self.age = age
    
    def getinfo(self):
        print(self.name , self.age)


ch1 = child('adi',12)

ch1.getinfo()
ch1.getname()
```

```python
# Multilevel level inheritance A child class inherits from a parent class, which in turn inherits from another parent class.
class parent :
    def  __init__(self,name):
        self.name = name

    def getname(self):
        print(self.name)


class child(parent):

    def  __init__(self,name,age):
        super().__init__(name)
        self.age = age
    
    def getinfo(self):
        print(self.name , self.age)



class person(child):

    def __init__(self,name,age,gender):
        super().__init__(name,age)
        self.gender = gender

    def info(self):
        print(self.name , self.age , self.gender)


p1 = person('adi',12,'female')

p1.getinfo()
p1.info()
```

```python
# Multiple A single child class inherits directly from more than one parent class.

class bike1 :
    def __init__(self,speed):
        self.speed = speed

    def getspeed(self):
        print(self.speed)    


class bike2 :
    def __init__(self,life):
        self.life = life
    
    def getlife(self):
        print(self.life)


class super_bike(bike1,bike2) :

    def __init__(self,spd , lif , avg) :
        bike1.__init__(self,spd)
        bike2.__init__(self,lif)
        self.avg = avg
    
    def info(self):
        print(self.life , self.speed , self.avg)

sb1 = super_bike(490,150,90)

sb1.info()
```

```python
# Hierarchical Inheritance Multiple child classes inherit from a single, shared parent class.


class parent:

    def __init__(self , gender):
        self.gender = gender

    def getgender(self):
        print(self.gender)

class dog(parent):
    def __init__(self,name , gender):
        super().__init__(gender)
        self.name = name

    def getinfo(self):
        print(self.name , self.gender)


class cat(parent):
    def __init__(self , name , gender):
        super().__init__(gender)
        self.name = name

    def getinfo(self):
        print(self.name , self.gender)

d1 = dog('solu','male')
c1 = cat('soli','female')

d1.getinfo()
c1.getinfo()
```

```python
class Parent:
    def __init__(self, gender):
        print("Parent constructor called!")
        self.gender = gender

    def getgender(self):
        print(self.gender)


class Dog(Parent):
    def __init__(self, name, gender):
        print("dog hu 1")
        Parent.__init__(self, gender)
        print("dog hu 2")
        self.name = name

    def getinfo(self):
        print(self.name, self.gender)


class Diff(Parent):
    def __init__(self, speed, gender):
        print("diff hu 1")
        Parent.__init__(self, gender)
        print("diff hu 2")
        self.speed = speed

    def getinfo(self):
        print(self.speed, self.gender)


class Mix(Dog, Diff):
    def __init__(self, name, speed, gender, power):
        print("mix hu 1")
        Dog.__init__(self, name, gender)
        print("mix hu 2")
        Diff.__init__(self, speed, gender)
        print("mix hu 3")
        self.power = power

    def info(self):
        print(self.name, self.gender, self.speed, self.power)


m1 = Mix('lalu', 120, 'male', 'invisible')

m1.info()
m1.getinfo()
m1.getgender()
```
